```python
import pandas as pd
from pathlib import Path
```

```python
DATA_DIR = Path("../data/raw")
OUTPUT_DIR = Path("../data/processed")

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
files = sorted(DATA_DIR.glob("solar_data_*.csv"))

print("Number of raw files:", len(files))

for file in files:
    print(file.name)
```

```python
for file in files:

    df = pd.read_csv(file)

    print("\n" + "=" * 60)
    print(file.name)
    print("=" * 60)

    print("Shape:", df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nFirst 3 dates:")
    print(df["Date"].head(3).tolist())

    print("\nLast 3 dates:")
    print(df["Date"].tail(3).tolist())
```

```python
standard_columns = [
    "Date",
    "Radio_Flux_10.7cm",
    "SESC_Sunspot_Number",
    "Sunspot_Area_10E-6_Hemis",
    "New_Regions",
    "Mean_Field",
    "Bkgd_XRay_Flux",
    "Flares_C",
    "Flares_M",
    "Flares_X",
    "Flares_S",
    "Flares_1",
    "Flares_2",
    "Flares_3"
]
```

```python
rename_map = {
    "RadioFlux10.7cm": "Radio_Flux_10.7cm",
    "SESCSunspotNumber": "SESC_Sunspot_Number",
    "SunspotArea": "Sunspot_Area_10E-6_Hemis",
    "NewRegions": "New_Regions",
    "MeanField": "Mean_Field",
    "BkgdXRayFlux": "Bkgd_XRay_Flux",

    "C": "Flares_C",
    "M": "Flares_M",
    "X": "Flares_X",
    "S": "Flares_S",

    "Optical1": "Flares_1",
    "Optical2": "Flares_2",
    "Optical3": "Flares_3",

    "Stanford_Mean_Field": "Mean_Field",

    "GOES15_Solar_Bkgd_Xray_Flux": "Bkgd_XRay_Flux",
    "GOES15_Solar_Bkgd_Flux": "Bkgd_XRay_Flux",

    "Xray_C": "Flares_C",
    "Xray_M": "Flares_M",
    "Xray_X": "Flares_X",

    "Optical_S": "Flares_S",
    "Optical_1": "Flares_1",
    "Optical_2": "Flares_2",
    "Optical_3": "Flares_3",

    "XRay_Flares_C": "Flares_C",
    "XRay_Flares_M": "Flares_M",
    "XRay_Flares_X": "Flares_X",
    "XRay_Flares_S": "Flares_S",

    "Optical_Flares_1": "Flares_1",
    "Optical_Flares_2": "Flares_2",
    "Optical_Flares_3": "Flares_3"
}
```

```python
all_years = []

for file in files:

    print("Reading:", file.name)

    df = pd.read_csv(file)

    # Rename columns
    df.rename(columns=rename_map, inplace=True)

    # Detect date format
    sample_date = str(df["Date"].dropna().iloc[0])

    if "-" in sample_date:

        df["Date"] = pd.to_datetime(
            df["Date"],
            format="%Y-%m-%d",
            errors="coerce"
        )

    else:

        df["Date"] = pd.to_datetime(
            df["Date"],
            format="%d/%m/%Y",
            errors="coerce"
        )

    # Add year
    df["Year"] = df["Date"].dt.year

    # Handle duplicate columns
    for column in standard_columns:

        matching = df.loc[:, df.columns == column]

        if matching.shape[1] > 1:
            df[column] = matching.bfill(axis=1).iloc[:, 0]

    # Remove duplicate column names
    df = df.loc[:, ~df.columns.duplicated()]

    # Add missing columns
    for column in standard_columns:

        if column not in df.columns:
            df[column] = pd.NA

    # Keep standard columns
    df = df[standard_columns + ["Year"]]

    all_years.append(df)

print("\nFiles processed:", len(all_years))
```

```python
combined = pd.concat(
    all_years,
    ignore_index=True
)

combined = (
    combined
    .sort_values("Date")
    .reset_index(drop=True)
)

print("Combined shape:", combined.shape)
```

```python
print("Date range:")
print(combined["Date"].min())
print("to")
print(combined["Date"].max())

print("\nInvalid dates:")
print(combined["Date"].isna().sum())

print("\nDuplicate dates:")
print(combined["Date"].duplicated().sum())
```

```python
print("Final columns:")

for i, column in enumerate(combined.columns, start=1):
    print(i, column)
```

```python
output_file = OUTPUT_DIR / "combined_raw.csv"

combined.to_csv(
    output_file,
    index=False
)

print("Saved:", output_file)
print("Shape:", combined.shape)
```
