```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(
    "../data/processed/combined_raw.csv",
    parse_dates=["Date"]
)

df = (
    df
    .sort_values("Date")
    .reset_index(drop=True)
)

print("Dataset shape:", df.shape)
```

```python
print("Date range:")
print(df["Date"].min(), "to", df["Date"].max())

print("\nInvalid dates:")
print(df["Date"].isna().sum())

print("\nDuplicate dates:")
print(df["Date"].duplicated().sum())
```

```python
print("Missing values:")
print(df.isna().sum())
```

```python
print("Total -999 values:")
print((df == -999).sum())

print("\nTotal '*' values:")
print((df == "*").sum())

print("\nTotal 'Unk' values:")
print((df == "Unk").sum())
```

```python
print("Mean_Field analysis")
print("=" * 50)

print("Total rows:", len(df))

print(
    "Number of -999 values:",
    (df["Mean_Field"] == -999).sum()
)

print(
    "Percentage of -999:",
    (df["Mean_Field"] == -999).mean() * 100,
    "%"
)

print("\nMost common values:")
print(df["Mean_Field"].value_counts().head(10))
```

```python
print("Bkgd_XRay_Flux analysis")
print("=" * 50)

print("Total rows:", len(df))

print(
    "Number of '*' values:",
    (df["Bkgd_XRay_Flux"] == "*").sum()
)

print(
    "Number of 'Unk' values:",
    (df["Bkgd_XRay_Flux"] == "Unk").sum()
)

print(
    "'*' or 'Unk' percentage:",
    df["Bkgd_XRay_Flux"]
    .isin(["*", "Unk"])
    .mean() * 100,
    "%"
)

print("\nMost common values:")
print(df["Bkgd_XRay_Flux"].value_counts().head(20))
```

```python
df = df.drop(
    columns=[
        "Mean_Field",
        "Bkgd_XRay_Flux"
    ]
)

print("Dropped:")
print("Mean_Field")
print("Bkgd_XRay_Flux")

print("\nRemaining columns:")
print(df.columns.tolist())
```

```python
flare_count_cols = [
    "Flares_C",
    "Flares_M",
    "Flares_X"
]
```

```python
for col in flare_count_cols:

    print(
        col,
        "(-1 count):",
        (df[col] == -1).sum()
    )
```

```python
df[flare_count_cols] = (
    df[flare_count_cols]
    .replace(-1, np.nan)
)
```

```python
print("NaN values after replacing -1:")

print(
    df[flare_count_cols].isna().sum()
)
```

```python
confirmed_outliers = pd.to_datetime([
    "2022-08-28",
    "2023-02-17",
    "2025-08-30"
])

print("Confirmed outliers:")
print(confirmed_outliers)
```

```python
df.loc[
    df["Date"].isin(confirmed_outliers),
    "Radio_Flux_10.7cm"
] = np.nan
```

```python
print(
    df[
        df["Date"].isin(confirmed_outliers)
    ][
        [
            "Date",
            "Radio_Flux_10.7cm"
        ]
    ]
)
```

```python
df["Radio_Flux_10.7cm"] = (
    df["Radio_Flux_10.7cm"]
    .interpolate()
)

print(
    "Remaining Radio Flux NaN:",
    df["Radio_Flux_10.7cm"].isna().sum()
)
```

```python
df[flare_count_cols] = (
    df[flare_count_cols]
    .ffill()
)
```

```python
print("Remaining NaNs:")
print(df.isna().sum())
```

```python
rolling_columns = [
    "Radio_Flux_10.7cm",
    "SESC_Sunspot_Number",
    "Sunspot_Area_10E-6_Hemis"
]
```

```python
for col in rolling_columns:

    for window in [3, 7, 27]:

        df[f"{col}_avg_{window}d"] = (
            df[col]
            .rolling(window)
            .mean()
        )
```

```python
rolling_features = [
    col
    for col in df.columns
    if "_avg_" in col
]

print("Rolling features:")

for col in rolling_features:
    print(col)
```

```python
print(
    df[rolling_features].isna().sum()
)
```

```python
df = (
    df
    .dropna(subset=rolling_features)
    .reset_index(drop=True)
)

print("Shape after rolling features:", df.shape)
```

```python
sample_idx = 50

manual_avg = (
    df["Radio_Flux_10.7cm"]
    .iloc[sample_idx-2:sample_idx+1]
    .mean()
)

computed_avg = (
    df["Radio_Flux_10.7cm_avg_3d"]
    .iloc[sample_idx]
)

print("Manual average:", manual_avg)
print("Computed average:", computed_avg)
```

```python
print("FINAL CLEANED DATASET")
print("=" * 50)

print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nRemaining NaNs:")
print(df.isna().sum())

print(
    "\nDates sorted:",
    df["Date"].is_monotonic_increasing
)
```

```python
output_file = "../data/processed/cleaned_data.csv"

df.to_csv(
    output_file,
    index=False
)

print("Saved:", output_file)
```
